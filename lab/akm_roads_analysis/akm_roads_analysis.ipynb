# Analysis of the OSM data of AKM roads

### Loading the Data

```python
import json

ROADS = None

with open('AKM_Roads.geojson', 'r') as f:
    ROADS = json.load(f)["features"]

print(len(ROADS))
```

## Visualising Data Health

```python
properties_keys = list(ROADS[0]["properties"].keys())

ROAD_PROPERTIES = []

for road in ROADS:
    properties = road["properties"]
    ROAD_PROPERTIES.append(properties)

len(ROAD_PROPERTIES), ROAD_PROPERTIES[0]
```

```python
import matplotlib.pyplot as plt

labels = list(ROAD_PROPERTIES[0].keys())

# 2. Count non-None values for each label across the entire list
counts = []
for label in labels:
    count = sum(1 for d in ROAD_PROPERTIES if d.get(label) is not None)
    counts.append(count)

# 3. Create the bar graph
plt.figure(figsize=(14, 6))
plt.bar(labels, counts, color='skyblue', edgecolor='navy')

plt.xticks(rotation=90)
plt.xlabel('Dictionary Keys')
plt.ylabel('Frequency (Not None)')
plt.title('Frequency of Populated Attributes in Dataset')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

````python
def get_reports_nearby(lon, lat, radius_m):

    QUERY = f"""SELECT *
    FROM reports
    WHERE ST_DWithin(
        location::geography,
        ST_SetSRID(ST_MakePoint({lon}, {lat}), 4326)::geography,
        {radius_m}
    );"""

    cur.execute(QUERY)
    rows = cur.fetchall()

    return rows

len(get_reports_nearby(88.63733761742397, 22.85270625962288, 100))

````

### Ward Table


```
CREATE TABLE wards (
    id SERIAL PRIMARY KEY,
    ward_no INTEGER,
    name TEXT,
    geometry GEOMETRY(POLYGON, 4326)
);
```

### Which ward contains this report

```
SELECT w.name
FROM wards w
JOIN reports r
ON ST_Contains(w.geometry, r.location)
WHERE r.id = 1;
```
